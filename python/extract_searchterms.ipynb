```python
# import libraries 

import re
from lxml import etree
```

```python
def get_text_from_path(path): 
    tree = etree.parse(path)
    root = tree.getroot()
    text = root.itertext()
    all_text = ''.join(root.itertext()) #could iter over it as well 
    all_words = re.findall(r"\w+", all_text)
    all_lower_words = [x.lower() for x in all_words]
    unique_words = set(all_lower_words)
    unique_words = list(unique_words)
    unique_words = sorted(unique_words)
    
    return unique_words
```
